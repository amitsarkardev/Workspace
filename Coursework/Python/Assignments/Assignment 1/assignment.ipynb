```python
# Build a sample student dataset
records = {
    "Name": ["Aarav", "Pooja", "Rohan", "Sneha", "Vikram"],
    "Department": ["CSE", "ECE", "CSE", "IT", "CSE"],
    "CGPA": [8.4, 7.9, 9.1, 8.8, 6.7],
    "Backlogs": [0, 1, 0, 0, 2]
}

df = pd.DataFrame(records)

# Filter students eligible for placements: CGPA >= 8.0 and zero backlogs
eligible = df[(df["CGPA"] >= 8.0) & (df["Backlogs"] == 0)]
eligible
```

```python
# 1. Save eligible students to a CSV file
eligible.to_csv("eligible_students.csv", index=False)

# 2. Read the CSV file back into a new DataFrame
loaded_df = pd.read_csv("eligible_students.csv")
loaded_df
```
