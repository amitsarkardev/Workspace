```python
# Master list used across all following cells
nums = [15, 42, 8, 93, 27, 42, 60, 15, 4, 88]
print(f"Master List loaded: {nums}")
```

```python
largest = nums[0]
smallest = nums[0]

for item in nums[1:]:
    if item > largest:
        largest = item
    if item < smallest:
        smallest = item

print(f"Largest element : {largest}")
print(f"Smallest element: {smallest}")
```

```python
total = 0
for item in nums:
    total += item

average = total / len(nums)

print(f"Sum    : {total}")
print(f"Average: {average:.2f}")
```

```python
even_count = 0
odd_count = 0

for item in nums:
    if item % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print(f"Even numbers count: {even_count}")
print(f"Odd numbers count : {odd_count}")
```

```python
unique_list = []

for item in nums:
    if item not in unique_list:
        unique_list.append(item)

print(f"Original list     : {nums}")
print(f"Without duplicates: {unique_list}")
```

```python
# Using slice step of -1
reversed_list = nums[::-1]

print(f"Original list: {nums}")
print(f"Reversed list: {reversed_list}")
```

```python
# Convert the list to a tuple first to work with tuple data
num_tuple = tuple(nums)

max_val = num_tuple[0]
min_val = num_tuple[0]

for item in num_tuple[1:]:
    if item > max_val:
        max_val = item
    if item < min_val:
        min_val = item

print(f"Working Tuple: {num_tuple}")
print(f"Maximum: {max_val}")
print(f"Minimum: {min_val}")
```

```python
even_items = []
odd_items = []

for item in nums:
    if item % 2 == 0:
        even_items.append(item)
    else:
        odd_items.append(item)

even_tuple = tuple(even_items)
odd_tuple = tuple(odd_items)

print(f"Even Tuple: {even_tuple}")
print(f"Odd Tuple : {odd_tuple}")
```

```python
converted_tuple = tuple(nums)
tuple_avg = sum(converted_tuple) / len(converted_tuple)

print(f"Converted Tuple: {converted_tuple}")
print(f"Average of Tuple: {tuple_avg:.2f}")
```

```python
first = second = float('-inf')

for item in nums:
    if item > first:
        second = first
        first = item
    elif item > second and item != first:
        second = item

print(f"List: {nums}")
print(f"Second largest element: {second}")
```

```python
# Use the master list nums alongside a second list
list2 = [10, 42, 50, 88, 99]

merged_unique = []
for item in nums + list2:
    if item not in merged_unique:
        merged_unique.append(item)

print(f"List 1 (Master): {nums}")
print(f"List 2         : {list2}")
print(f"Merged Unique  : {merged_unique}")
```
