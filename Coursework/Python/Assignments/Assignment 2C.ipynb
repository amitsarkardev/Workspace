```python
student_records = {
    "Amit": 82,
    "Rohan": 55,
    "Pooja": 91,
    "Sneha": 48,
    "Vikram": 74,
    "Ananya": 60
}

print("Students who scored more than 60:")
for student, marks in student_records.items():
    if marks > 60:
        print(f"- {student}: {marks} marks")
```

```python
employee_salaries = {
    "Ramesh": 55000,
    "Priya": 82000,
    "Deepak": 67000,
    "Tanvi": 95000,
    "Suresh": 48000
}

highest_earner = None
max_salary = float('-inf')

for name, salary in employee_salaries.items():
    if salary > max_salary:
        max_salary = salary
        highest_earner = name

print(f"Employee with highest salary: {highest_earner} (₹{max_salary:,.2f})")
```

```python
sentence = "python is simple and python is powerful and python is versatile"

words = sentence.split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print("Word Frequencies:")
for word, count in frequency.items():
    print(f"'{word}': {count}")
```

```python
squares_dict = {x: x**2 for x in range(1, 11)}

print("Dictionary of numbers (1 to 10) and their squares:")
for num, sq in squares_dict.items():
    print(f"{num:2d} -> {sq:3d}")
```

```python
student_data = {"name": "Amit", "department": "CSE", "semester": 3}
print(f"Initial Dictionary: {student_data}")

# 1. Add a new key-value pair
student_data["college"] = "TMSL"
print(f"After Adding 'college': {student_data}")

# 2. Update an existing key
student_data["semester"] = 4
print(f"After Updating 'semester': {student_data}")

# 3. Delete an element using pop()
removed_dept = student_data.pop("department")
print(f"After Deleting 'department' (removed '{removed_dept}'): {student_data}")
```

```python
# Structure: {product_name: {"quantity": qty, "price": unit_price}}
products = {
    "Laptop": {"quantity": 4, "price": 55000.0},
    "Mouse": {"quantity": 15, "price": 450.0},
    "Keyboard": {"quantity": 8, "price": 1200.0},
    "Monitor": {"quantity": 5, "price": 16500.0}
}

total_inventory_cost = 0.0

print(f"{'Product':<12} | {'Qty':<6} | {'Price/Unit':<12} | {'Subtotal':<12}")
print("-" * 50)

for product, info in products.items():
    subtotal = info["quantity"] * info["price"]
    total_inventory_cost += subtotal
    print(f"{product:<12} | {info['quantity']:<6} | ₹{info['price']:<10.2f} | ₹{subtotal:<10.2f}")

print("-" * 50)
print(f"Total Cost of All Products: ₹{total_inventory_cost:,.2f}")
```
