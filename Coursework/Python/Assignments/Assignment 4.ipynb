```python
import numpy as np

# Master 3x3 matrices used across the following cells
A = np.array([
    [10, 25, 14],
    [30, 42, 55],
    [18, 90,  7]
])

B = np.array([
    [2, 5, 1],
    [3, 4, 6],
    [7, 2, 8]
])

print("Master Matrix A (3x3):")
print(A)
print("\nMaster Matrix B (3x3):")
print(B)
```

```python
# Statistical calculations on master matrix A
total_sum = np.sum(A)
mean_val = np.mean(A)
median_val = np.median(A)
std_dev = np.std(A)

print(f"Sum               : {total_sum}")
print(f"Mean              : {mean_val:.2f}")
print(f"Median            : {median_val}")
print(f"Standard Deviation: {std_dev:.2f}")
```

```python
# Transpose of master matrix A
A_transposed = A.T

print("Original Matrix A:")
print(A)
print("\nTransposed Matrix A:")
print(A_transposed)
```

```python
# Boolean masking on master matrix A to extract multiples of 5
divisible_by_5 = A[A % 5 == 0]

print("Elements in Matrix A divisible by 5:")
print(divisible_by_5)
```

```python
# Element-wise arithmetic operations between master matrices A and B
print("Addition (A + B):")
print(A + B)

print("\nSubtraction (A - B):")
print(A - B)

print("\nElement-wise Multiplication (A * B):")
print(A * B)

print("\nElement-wise Division (A / B):")
print(np.round(A / B, 2))
```

```python
# Generate an array of 10 random integers between 1 and 100
np.random.seed(42)  # Fixed seed for reproducible output
rand_arr = np.random.randint(1, 101, size=10)

print(f"Generated Random Array: {rand_arr}")
print(f"Maximum Value: {np.max(rand_arr)}")
print(f"Minimum Value: {np.min(rand_arr)}")
```

```python
# axis=1 sums across columns (row-wise), axis=0 sums across rows (column-wise)
row_sums = np.sum(A, axis=1)
col_sums = np.sum(A, axis=0)

print("Matrix A:")
print(A)
print(f"\nRow-wise Sums   : {row_sums}")
print(f"Column-wise Sums: {col_sums}")
```

```python
# Flatten matrix A, get unique values, and sort them
unique_sorted = np.sort(np.unique(A))

if len(unique_sorted) < 2:
    print("Array does not contain a second-largest element.")
else:
    second_largest = unique_sorted[-2]
    print(f"Flattened Unique Elements: {unique_sorted}")
    print(f"Second-Largest Element   : {second_largest}")
```

```python
# Matrix dot product / matrix multiplication of A and B (3x3 @ 3x3)
matrix_product = np.matmul(A, B)  # Or: A @ B

print("Matrix Multiplication (A @ B):")
print(matrix_product)
```
