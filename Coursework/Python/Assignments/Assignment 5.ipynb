```python
# 11. Calculate the cumulative sum of a NumPy array and store in a new column
import pandas as pd
import numpy as np

df = pd.read_csv("employee_data.csv").head(10) # Using top 10 for visibility
df["Salary"] = df["Salary"].fillna(0) # Handle NaN so cumulative sum works

# Step 1: Extract the Pandas column as a pure NumPy array
salary_array = df["Salary"].to_numpy()

# Step 2: Use NumPy's cumulative sum function
cumulative_array = np.cumsum(salary_array)

# Step 3: Assign the resulting NumPy array back into the DataFrame as a new column
df["Running_Total_Expense"] = cumulative_array

print("--- Cumulative Sum of Salaries ---")
print(df[["Emp_ID", "Salary", "Running_Total_Expense"]])
```

```python
# 12. Group a Pandas DataFrame by a column and calculate the mean of another column
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Use groupby() to group rows by the unique values in 'Department'
# Step 2: Select the 'Salary' column from those groups
# Step 3: Apply the mean() aggregation function
average_salary_by_dept = df.groupby("Department")["Salary"].mean()

# Step 4: Format the output to 2 decimal places for currency
average_salary_by_dept = average_salary_by_dept.round(2)

print("--- Average Salary by Department ---")
print(average_salary_by_dept.reset_index())
```

```python
# 13. Reshape a Pandas DataFrame using the pivot_table function
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Use pivot_table to create a 2D matrix summary of the data
# index: What becomes the rows (Department)
# columns: What becomes the columns (Role)
# values: What data populates the cells (Salary)
# aggfunc: How to combine multiple entries in the same cell (mean/average)
pivot = pd.pivot_table(
    df, 
    values="Salary", 
    index="Department", 
    columns="Role", 
    aggfunc="mean"
)

# Step 2: Fill NaN values (where a department doesn't have a specific role) with 0
pivot_clean = pivot.fillna(0).round(2)

print("--- Pivot Table: Average Salary by Dept and Role ---")
print(pivot_clean)
```

```python
# 14. Replace missing values with the mean of the column
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Identify missing values before operation
print("Missing values in Salary before filling:", df["Salary"].isna().sum())
print("Row with missing salary:\n", df[df["Salary"].isna()][["Emp_ID", "Salary"]])

# Step 2: Calculate the mean of the Salary column (Pandas automatically ignores NaNs here)
salary_mean = df["Salary"].mean()
print(f"\nCalculated Mean Salary: ₹{salary_mean:.2f}")

# Step 3: Use the fillna() method to replace all NaN values with the calculated mean
df["Salary"] = df["Salary"].fillna(salary_mean)

print("\nMissing values in Salary after filling:", df["Salary"].isna().sum())
print("Previously missing row updated:\n", df[df["Emp_ID"] == 105][["Emp_ID", "Salary"]])
```

```python
# 16. Normalize a numerical column in a Pandas DataFrame
import pandas as pd

df = pd.read_csv("employee_data.csv")
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Step 1: Find the minimum and maximum values of the column
min_val = df["Salary"].min()
max_val = df["Salary"].max()

# Step 2: Apply Min-Max Normalization formula: (x - min) / (max - min)
# This scales all salaries to a strict range between 0.0 and 1.0
df["Normalized_Salary"] = (df["Salary"] - min_val) / (max_val - min_val)

print("--- Min-Max Normalized Salaries ---")
print(df[["Emp_ID", "Salary", "Normalized_Salary"]].head(10))
```

```python
# 17. Remove duplicate rows from a Pandas DataFrame
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Check the initial length of the DataFrame
initial_count = len(df)
print(f"Initial number of rows: {initial_count}")

# Step 2: Use drop_duplicates() to remove identical rows
# Keep='first' ensures the original row remains, while subsequent identical copies are deleted
df_cleaned = df.drop_duplicates(keep="first")

# Step 3: Check the final length
final_count = len(df_cleaned)
print(f"Number of rows after removing duplicates: {final_count}")
print(f"Total duplicates removed: {initial_count - final_count}")

print("\n--- Bottom of Cleaned DataFrame ---")
print(df_cleaned.tail(3))
```

```python
# 19. Apply a custom function to each element in a Pandas DataFrame
import pandas as pd
import numpy as np

df = pd.read_csv("employee_data.csv")

# Step 1: Select ONLY numerical columns, because string data will crash math operations
num_df = df.select_dtypes(include=[np.number]).dropna()

# Step 2: Define a custom function. Here we add 1000 to the value if it's over 100, 
# otherwise we leave it alone (protecting smaller values like Experience/ID)
def custom_logic(val):
    if val > 1000:
        return val + 1000
    return val

# Step 3: Use the map() function to apply this to EVERY individual cell in the numeric DataFrame
modified_df = num_df.map(custom_logic)

print("--- Original Numeric Data (Head) ---")
print(num_df[["Emp_ID", "Experience", "Salary"]].head())

print("\n--- Modified Numeric Data (Salary increased by 1000) ---")
print(modified_df[["Emp_ID", "Experience", "Salary"]].head())
```

```python
# 22. Merge two DataFrames based on multiple columns
import pandas as pd

df = pd.read_csv("employee_data.csv").head(10)

# Step 1: Create a secondary DataFrame defining shift timings based on BOTH Department AND Role
shift_data = {
    "Department": ["IT", "IT", "Finance", "HR"],
    "Role": ["Dev", "Analyst", "Manager", "Manager"],
    "Shift_Timing": ["Night", "Day", "Day", "Day"]
}
df_shifts = pd.DataFrame(shift_data)

# Step 2: Merge using a list of column names for the 'on' parameter
# An employee must match BOTH Department and Role to get the shift timing assigned
complex_merge = pd.merge(df, df_shifts, on=["Department", "Role"], how="inner")

print("--- Merged on Department AND Role ---")
print(complex_merge[["Emp_ID", "Department", "Role", "Shift_Timing"]])
```

```python
# 23. Aggregate data in a DataFrame by multiple functions
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Group the data by Department
grouped = df.groupby("Department")

# Step 2: Select the 'Salary' column and use the .agg() method
# Pass a list of strings representing the statistical functions to calculate simultaneously
aggregation_result = grouped["Salary"].agg(["count", "mean", "min", "max", "sum"])

# Step 3: Format the output for readability
aggregation_result = aggregation_result.round(0)

print("--- Multi-Function Salary Aggregation by Department ---")
print(aggregation_result)
```

```python
# 24. Extract the date and time components from a DateTime column
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Ensure the Timestamp column is explicitly cast to Pandas datetime format
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Step 2: Use the .dt accessor to extract specific components into new columns
df["Extracted_Year"] = df["Timestamp"].dt.year
df["Extracted_Month"] = df["Timestamp"].dt.month_name()
df["Extracted_Day"] = df["Timestamp"].dt.day_name()
df["Extracted_Time"] = df["Timestamp"].dt.time

print("--- Extracted Date and Time Components ---")
print(df[["Timestamp", "Extracted_Month", "Extracted_Day", "Extracted_Time"]].head(5))
```

```python
# 25. Resample time-series data in a DataFrame
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Convert Timestamp to datetime objects
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Step 2: Set the Timestamp column as the actual index of the DataFrame
# Resampling strictly requires a DatetimeIndex
df_time = df.set_index("Timestamp")

# Step 3: Use resample() to change the frequency of the data
# "W" means Weekly. We are calculating the average salary recorded per week
# (Based on the daily rows we generated in Cell 0)
weekly_resampled = df_time["Salary"].resample("W").mean().round(2)

print("--- Weekly Resampled Average Salaries ---")
print(weekly_resampled)
```

```python
# 26. Perform a rolling calculation on a numerical column in a DataFrame
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Sort by Timestamp to ensure time-series order is chronological
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
df = df.sort_values(by="Timestamp").reset_index(drop=True)

# Step 2: Use the rolling() function on the Salary column with a window of 3
# This calculates statistics across the current row and the 2 preceding rows
# Step 3: Apply the mean() function to calculate the 3-day moving average
df["3_Day_Moving_Avg"] = df["Salary"].rolling(window=3).mean().round(2)

print("--- Rolling Calculation (3-Day Moving Average) ---")
# Displaying from index 1 to 6 to show the first NaN outputs and the subsequent values
print(df[["Timestamp", "Salary", "3_Day_Moving_Avg"]].head(7))
```

```python
# 27. Perform a cross-tabulation between two columns in a DataFrame
import pandas as pd

df = pd.read_csv("employee_data.csv")

# Step 1: Use pd.crosstab() to compute a simple cross-tabulation of two factors
# This counts the frequency of occurrences for each combination of Department and Role
cross_tab = pd.crosstab(
    index=df["Department"],   # Rows
    columns=df["Role"],       # Columns
    margins=True,             # Adds Row/Column subtotals ('All')
    margins_name="Total"
)

print("--- Cross-Tabulation: Department vs Role ---")
print(cross_tab)
```
