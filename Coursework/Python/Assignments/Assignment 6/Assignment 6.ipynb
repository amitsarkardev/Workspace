```python
# Assignment 1: Student Marks Analysis
import numpy as np
import pandas as pd

# Step 1: Create a NumPy array containing the marks of 5 students
marks_array = np.array([78, 85, 62, 90, 74])
print("--- 1. Original NumPy Marks Array ---")
print(marks_array)

# Step 2: Calculate the maximum, minimum, and average marks using NumPy functions
max_marks = np.max(marks_array)
min_marks = np.min(marks_array)
avg_marks = np.mean(marks_array)

print("\n--- 2. Statistical Summary (NumPy) ---")
print(f"Maximum Marks : {max_marks}")
print(f"Minimum Marks : {min_marks}")
print(f"Average Marks : {avg_marks:.2f}")

# Step 3: Create a Pandas DataFrame with columns 'Name' and 'Marks'
student_names = ["Amit", "Rohan", "Priya", "Sneha", "Vikram"]
df_students = pd.DataFrame({
    "Name": student_names,
    "Marks": marks_array
})

print("\n--- 3. Complete Student Marks DataFrame ---")
print(df_students)

# Step 4: Filter and display students who scored strictly more than 75 marks
# Boolean indexing creates a mask where Marks > 75 is True
high_scorers = df_students[df_students["Marks"] > 75]
print("\n--- 4. Students Who Scored More Than 75 Marks ---")
print(high_scorers)

# Step 5: Find and display the student with the highest marks
# idxmax() returns the row index where the 'Marks' column has its maximum value
topper_index = df_students["Marks"].idxmax()
topper_details = df_students.loc[topper_index]

print("\n--- 5. Student With the Highest Marks ---")
print(topper_details)
```

```python
# Assignment 2: Product Sales Analysis
import numpy as np
import pandas as pd

# Step 1: Create a NumPy array containing the given sales values
sales_array = np.array([1200, 1500, 900, 1800, 2000])
print("--- 1. NumPy Sales Array ---")
print(sales_array)

# Step 2: Find the total and average sales using NumPy
total_sales = np.sum(sales_array)
average_sales = np.mean(sales_array)

print("\n--- 2. Sales Metrics (NumPy) ---")
print(f"Total Sales   : ₹{total_sales}")
print(f"Average Sales : ₹{average_sales:.2f}")

# Step 3: Create a Pandas DataFrame with columns 'Product' and 'Sales'
product_names = ["Laptop", "Monitor", "Keyboard", "Smartphone", "Graphics Card"]
df_sales = pd.DataFrame({
    "Product": product_names,
    "Sales": sales_array
})

print("\n--- 3. Complete Product Sales DataFrame ---")
print(df_sales)

# Step 4: Display products having sales greater than 1400
# This filters rows where the 'Sales' column value exceeds 1400
high_performing_products = df_sales[df_sales["Sales"] > 1400]
print("\n--- 4. Products With Sales Greater Than 1400 ---")
print(high_performing_products)

# Step 5: Find and display the product with the highest sales
highest_sales_idx = df_sales["Sales"].idxmax()
best_selling_product = df_sales.loc[highest_sales_idx]

print("\n--- 5. Product With the Highest Sales ---")
print(best_selling_product)
```

```python
# Assignment 3: Employee Salary Analysis
import numpy as np
import pandas as pd

# Step 1: Create a NumPy array containing salaries of 5 employees (in INR)
salaries_array = np.array([45000, 62000, 38000, 85000, 54000])
print("--- 1. Employee Salaries Array ---")
print(salaries_array)

# Step 2: Find the maximum, minimum, and average salary using NumPy
max_salary = np.max(salaries_array)
min_salary = np.min(salaries_array)
avg_salary = np.mean(salaries_array)

print("\n--- 2. Salary Statistics (NumPy) ---")
print(f"Maximum Salary : ₹{max_salary}")
print(f"Minimum Salary : ₹{min_salary}")
print(f"Average Salary : ₹{avg_salary:.2f}")

# Step 3: Create a Pandas DataFrame with 'Employee Name' and 'Salary'
employee_names = ["Amit", "Rohan", "Priya", "Sneha", "Vikram"]
df_employees = pd.DataFrame({
    "Employee Name": employee_names,
    "Salary": salaries_array
})

print("\n--- 3. Employee Salary DataFrame ---")
print(df_employees)

# Step 4: Display employees whose salary is above the calculated average salary
above_avg_employees = df_employees[df_employees["Salary"] > avg_salary]
print(f"\n--- 4. Employees Earning Above Average (> ₹{avg_salary:.2f}) ---")
print(above_avg_employees)

# Step 5: Sort the DataFrame by salary in descending order (highest to lowest)
df_sorted_salaries = df_employees.sort_values(by="Salary", ascending=False)
print("\n--- 5. Employees Sorted by Salary (Descending Order) ---")
print(df_sorted_salaries)
```

```python
# Assignment 4: Hospital Management System
import numpy as np
import pandas as pd

# Step 1: Create a NumPy array containing the ages of 5 patients and display it
patient_ages = np.array([28, 56, 42, 64, 35])
print("--- 1. Patient Ages Array ---")
print(patient_ages)

# Step 2: Calculate the average, maximum, and minimum age using NumPy
avg_age = np.mean(patient_ages)
max_age = np.max(patient_ages)
min_age = np.min(patient_ages)

print("\n--- 2. Patient Age Statistics (NumPy) ---")
print(f"Average Age : {avg_age:.1f} years")
print(f"Maximum Age : {max_age} years")
print(f"Minimum Age : {min_age} years")

# Step 3: Create a Pandas DataFrame with Patient ID, Name, Age, Disease, and Treatment Cost
df_hospital = pd.DataFrame({
    "Patient ID": ["P101", "P102", "P103", "P104", "P105"],
    "Patient Name": ["Amit", "Rohan", "Priya", "Sneha", "Vikram"],
    "Age": patient_ages,
    "Disease": ["Viral Fever", "Cardiac Care", "Orthopaedic", "Hypertension", "Seasonal Flu"],
    "Treatment Cost": [4500, 32000, 15000, 18500, 3000]
})

print("\n--- 3. Initial Hospital Patient Records ---")
print(df_hospital)

# Step 4: Display all patients whose treatment cost is greater than ₹10,000
high_cost_patients = df_hospital[df_hospital["Treatment Cost"] > 10000]
print("\n--- 4. Patients With Treatment Cost > ₹10,000 ---")
print(high_cost_patients)

# Step 5: Find the average treatment cost using Pandas .mean() method
avg_treatment_cost = df_hospital["Treatment Cost"].mean()
print(f"\n--- 5. Average Treatment Cost (Pandas): ₹{avg_treatment_cost:,.2f} ---")

# Step 6: Add a new column named 'Status' with values "Admitted" or "Discharged"
df_hospital["Status"] = ["Discharged", "Admitted", "Admitted", "Admitted", "Discharged"]
print("\n--- 6. Updated DataFrame With 'Status' Column ---")
print(df_hospital)

# Step 7: Display patients whose age is above 50 years
senior_patients = df_hospital[df_hospital["Age"] > 50]
print("\n--- 7. Patients Aged Above 50 Years ---")
print(senior_patients)

# Step 8: Find and display the patient with the highest treatment cost
max_cost_idx = df_hospital["Treatment Cost"].idxmax()
highest_cost_patient = df_hospital.loc[max_cost_idx]
print("\n--- 8. Patient With the Highest Treatment Cost ---")
print(highest_cost_patient)

# Step 9: Sort the patient records by treatment cost in descending order
df_hospital_sorted = df_hospital.sort_values(by="Treatment Cost", ascending=False)
print("\n--- 9. Records Sorted by Treatment Cost (Descending) ---")
print(df_hospital_sorted)

# Step 10: Save the final sorted hospital records as 'hospital_records.csv'
df_hospital_sorted.to_csv("hospital_records.csv", index=False)
print("\n--- 10. File Export ---")
print("Successfully saved final records to 'hospital_records.csv'.")
```

```python
# Assignment 5: Movie Booking System
import numpy as np
import pandas as pd

# Step 1: Create a NumPy array containing the number of tickets sold for 5 movies
tickets_sold = np.array([150, 85, 220, 95, 180])
print("--- 1. Tickets Sold Array ---")
print(tickets_sold)

# Step 2: Calculate the total, average, maximum, and minimum tickets sold using NumPy
total_tickets = np.sum(tickets_sold)
avg_tickets = np.mean(tickets_sold)
max_tickets = np.max(tickets_sold)
min_tickets = np.min(tickets_sold)

print("\n--- 2. Ticket Sales Statistics (NumPy) ---")
print(f"Total Tickets Sold   : {total_tickets}")
print(f"Average Tickets Sold : {avg_tickets:.1f}")
print(f"Maximum Tickets Sold : {max_tickets}")
print(f"Minimum Tickets Sold : {min_tickets}")

# Step 3: Create a Pandas DataFrame with Movie Name, Genre, Ticket Price, and Tickets Sold
df_movies = pd.DataFrame({
    "Movie Name": ["Inception", "Avatar", "Interstellar", "Dune", "Oppenheimer"],
    "Genre": ["Sci-Fi", "Action", "Sci-Fi", "Adventure", "Biography"],
    "Ticket Price": [250, 300, 220, 280, 320],
    "Tickets Sold": tickets_sold
})

print("\n--- 3. Initial Movie Booking DataFrame ---")
print(df_movies)

# Step 4: Add a new column named 'Revenue' using Revenue = Ticket Price * Tickets Sold
df_movies["Revenue"] = df_movies["Ticket Price"] * df_movies["Tickets Sold"]
print("\n--- 4. Movie DataFrame With Calculated Revenue ---")
print(df_movies)

# Step 5: Display movies that sold more than 100 tickets
popular_movies = df_movies[df_movies["Tickets Sold"] > 100]
print("\n--- 5. Movies That Sold More Than 100 Tickets ---")
print(popular_movies)

# Step 6: Find the movie with the highest revenue
highest_rev_idx = df_movies["Revenue"].idxmax()
top_earning_movie = df_movies.loc[highest_rev_idx]
print("\n--- 6. Movie With the Highest Revenue ---")
print(top_earning_movie)

# Step 7: Calculate the total revenue generated by all movies
total_cinema_revenue = df_movies["Revenue"].sum()
print(f"\n--- 7. Total Revenue Generated by All Movies: ₹{total_cinema_revenue:,} ---")

# Step 8: Sort the movies according to revenue in descending order
df_movies_sorted = df_movies.sort_values(by="Revenue", ascending=False)
print("\n--- 8. Movies Sorted by Revenue (Descending) ---")
print(df_movies_sorted)

# Step 9: Find the average ticket price using Pandas
avg_ticket_price = df_movies["Ticket Price"].mean()
print(f"\n--- 9. Average Ticket Price: ₹{avg_ticket_price:.2f} ---")

# Step 10: Save the final movie booking report as 'movie_booking.csv'
df_movies_sorted.to_csv("movie_booking.csv", index=False)
print("\n--- 10. File Export ---")
print("Successfully saved final report to 'movie_booking.csv'.")
```

```python
# Assignment 6: Traffic Monitoring System
import numpy as np
import pandas as pd

# Step 1: Create a NumPy array containing the number of vehicles recorded on 5 roads
# Note: These values match the sum of Cars + Buses + Bikes in Step 3
vehicles_recorded = np.array([620, 460, 800, 385, 530])
print("--- 1. Vehicles Recorded Array ---")
print(vehicles_recorded)

# Step 2: Find the total, average, maximum, and minimum number of vehicles using NumPy
total_vehicles_np = np.sum(vehicles_recorded)
avg_vehicles_np = np.mean(vehicles_recorded)
max_vehicles_np = np.max(vehicles_recorded)
min_vehicles_np = np.min(vehicles_recorded)

print("\n--- 2. Overall Vehicle Statistics (NumPy) ---")
print(f"Total Vehicles   : {total_vehicles_np}")
print(f"Average Vehicles : {avg_vehicles_np:.1f}")
print(f"Maximum Vehicles : {max_vehicles_np}")
print(f"Minimum Vehicles : {min_vehicles_np}")

# Step 3: Create a Pandas DataFrame containing Road Name, Location, Cars, Buses, and Bikes
df_traffic = pd.DataFrame({
    "Road Name": ["MG Road", "Park Street", "EM Bypass", "VIP Road", "BT Road"],
    "Location": ["Central", "South", "East", "North-East", "North"],
    "Cars": [300, 220, 400, 180, 250],
    "Buses": [70, 40, 100, 35, 60],
    "Bikes": [250, 200, 300, 170, 220]
})

print("\n--- 3. Initial Traffic Monitoring DataFrame ---")
print(df_traffic)

# Step 4: Add a column named 'Total Vehicles' by adding Cars, Buses, and Bikes
df_traffic["Total Vehicles"] = df_traffic["Cars"] + df_traffic["Buses"] + df_traffic["Bikes"]
print("\n--- 4. DataFrame With 'Total Vehicles' Column ---")
print(df_traffic)

# Step 5: Display roads where the total number of vehicles is greater than 500
busy_roads = df_traffic[df_traffic["Total Vehicles"] > 500]
print("\n--- 5. Roads With Total Vehicles > 500 ---")
print(busy_roads)

# Step 6: Find the road with the highest traffic
highest_traffic_idx = df_traffic["Total Vehicles"].idxmax()
highest_traffic_road = df_traffic.loc[highest_traffic_idx]
print("\n--- 6. Road With the Highest Traffic ---")
print(highest_traffic_road)

# Step 7: Calculate the average number of cars recorded on all roads
avg_cars = df_traffic["Cars"].mean()
print(f"\n--- 7. Average Number of Cars Across All Roads: {avg_cars:.1f} ---")

# Step 8: Sort the roads by 'Total Vehicles' in descending order
df_traffic_sorted = df_traffic.sort_values(by="Total Vehicles", ascending=False)
print("\n--- 8. Roads Sorted by Total Vehicles (Descending) ---")
print(df_traffic_sorted)

# Step 9: Identify the road with the lowest traffic using idxmin()
lowest_traffic_idx = df_traffic["Total Vehicles"].idxmin()
lowest_traffic_road = df_traffic.loc[lowest_traffic_idx]
print("\n--- 9. Road With the Lowest Traffic ---")
print(lowest_traffic_road)

# Step 10: Save the final traffic report as 'traffic_monitoring.csv'
df_traffic_sorted.to_csv("traffic_monitoring.csv", index=False)
print("\n--- 10. File Export ---")
print("Successfully saved final traffic report to 'traffic_monitoring.csv'.")
```
