```python
text = input("Enter a string: ")

if len(text) < 3:
    print("String must have at least 3 characters.")
else:
    third_last_char = text[-3]
    if third_last_char.lower() in 'aeiou':
        print(f"The 3rd last character '{third_last_char}' is a vowel.")
    else:
        print(f"The 3rd last character '{third_last_char}' is not a vowel.")
```

```python
text = input("Enter a string: ").strip()

if not text:
    print("Empty string provided.")
elif text[0].lower() in 'aeiou':
    print(f"The string starts with a vowel ('{text[0]}').")
else:
    print(f"The string does not start with a vowel ('{text[0]}').")
```

```python
print("Enter marks obtained in 5 subjects (out of 100 each):")
m1 = float(input("Subject 1: "))
m2 = float(input("Subject 2: "))
m3 = float(input("Subject 3: "))
m4 = float(input("Subject 4: "))
m5 = float(input("Subject 5: "))

total_marks = m1 + m2 + m3 + m4 + m5
percentage = (total_marks / 500) * 100

if percentage >= 90:
    grade = "O (Outstanding)"
elif percentage >= 80:
    grade = "E (Excellent)"
elif percentage >= 70:
    grade = "A (Very Good)"
elif percentage >= 60:
    grade = "B (Good)"
elif percentage >= 50:
    grade = "C (Fair)"
elif percentage >= 40:
    grade = "D (Pass)"
else:
    grade = "F (Fail)"

print(f"\nTotal Marks: {total_marks:.2f}/500")
print(f"Percentage : {percentage:.2f}%")
print(f"Grade      : {grade}")
```

```python
import math

a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

# Check triangle inequality validity
if (a + b > c) and (a + c > b) and (b + c > a):
    s = (a + b + c) / 2
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))
    print(f"The area of the triangle is: {area:.2f} sq units")
else:
    print("The given side lengths do not form a valid triangle.")
```

```python
text = input("Enter a string: ")

# Method 1: Using built-in len()
count_builtin = len(text)

# Method 2: Iterative count
count_loop = 0
for _ in text:
    count_loop += 1

print(f"Total count of characters/elements in the string: {count_builtin}")
```

```python
side1 = float(input("Enter length of first side: "))
side2 = float(input("Enter length of second side: "))
side3 = float(input("Enter length of third side: "))

if side1 <= 0 or side2 <= 0 or side3 <= 0:
    print("Side lengths must be positive numbers.")
elif side1 == side2 == side3:
    print("It is an Equilateral triangle.")
else:
    print("It is NOT an Equilateral triangle.")
```

```python
num = int(input("Enter an integer: "))
abs_num = abs(num)

if abs_num < 10:
    print("Number must contain at least 2 digits.")
else:
    # (abs_num // 10) strips the last digit; % 10 extracts the new last digit (2nd last)
    second_last_digit = (abs_num // 10) % 10
    
    if second_last_digit % 3 == 0:
        print(f"The 2nd last digit is {second_last_digit}, which is divisible by 3.")
    else:
        print(f"The 2nd last digit is {second_last_digit}, which is NOT divisible by 3.")
```
