```python
def check_even_odd(n: int) -> str:
    if n % 2 == 0:
        return f"{n} is Even"
    return f"{n} is Odd"

# Test
print(check_even_odd(14))
print(check_even_odd(7))
```

```python
def find_largest(a: float, b: float, c: float) -> float:
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

# Test
print("Largest:", find_largest(25, 78, 42))
```

```python
def calculate_factorial(n: int) -> int:
    if n < 0:
        raise ValueError("Factorial is not defined for negative integers.")
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

# Test
num = 5
print(f"Factorial of {num}:", calculate_factorial(num))
```

```python
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Test
val = 29
print(f"Is {val} prime?:", is_prime(val))
print(f"Is 15 prime?:", is_prime(15))
```

```python
def is_palindrome(value) -> bool:
    s = str(value)
    return s == s[::-1]

# Test
print("madam is palindrome:", is_palindrome("madam"))
print("12321 is palindrome:", is_palindrome(12321))
print("python is palindrome:", is_palindrome("python"))
```

```python
def generate_fibonacci(n: int) -> list:
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    
    series = [0, 1]
    while len(series) < n:
        series.append(series[-1] + series[-2])
    return series

# Test
print("First 10 Fibonacci terms:", generate_fibonacci(10))
```

```python
def list_summary(lst: list):
    if not lst:
        raise ValueError("List cannot be empty.")
    
    total = sum(lst)
    maximum = max(lst)
    minimum = min(lst)
    return total, maximum, minimum

# Test
numbers = [12, 45, 2, 89, 34, 6]
s, high, low = list_summary(numbers)
print(f"Sum: {s}, Max: {high}, Min: {low}")
```

```python
def count_vowels(text: str) -> int:
    vowels = set("aeiouAEIOU")
    return sum(1 for char in text if char in vowels)

# Test
sample_str = "Artificial Intelligence and Machine Learning"
print(f"Number of vowels in '{sample_str}':", count_vowels(sample_str))
```

```python
class InsufficientBalanceError(Exception):
    """Raised when withdrawal amount exceeds available balance."""
    pass

def withdraw(balance: float, amount: float) -> float:
    if amount > balance:
        raise InsufficientBalanceError(
            f"Withdrawal failed: Requested ₹{amount}, Available balance: ₹{balance}"
        )
    balance -= amount
    print(f"Withdrawal of ₹{amount} successful. Remaining balance: ₹{balance}")
    return balance

# Test
current_balance = 5000.0
try:
    current_balance = withdraw(current_balance, 2000.0)
    current_balance = withdraw(current_balance, 4500.0)
except InsufficientBalanceError as err:
    print("Caught Exception:", err)
```

```python
def analyze_numbers(numbers: list):
    if not numbers:
        raise ValueError("List cannot be empty.")
    
    total = sum(numbers)
    average = total / len(numbers)
    maximum = max(numbers)
    minimum = min(numbers)
    even_count = sum(1 for x in numbers if x % 2 == 0)
    
    return {
        "sum": total,
        "average": average,
        "maximum": maximum,
        "minimum": minimum,
        "even_elements_count": even_count
    }

# Test
data = [14, 7, 22, 9, 30, 18, 5, 41]
stats = analyze_numbers(data)
for k, v in stats.items():
    print(f"{k}: {v}")
```

```python
class Student:
    def __init__(self, name: str, roll_number: str, marks: list):
        self.name = name
        self.roll_number = roll_number
        self.marks = self._validate_marks(marks)

    def _validate_marks(self, marks: list) -> list:
        if len(marks) != 5:
            raise ValueError("Exactly 5 subject marks must be provided.")
        for m in marks:
            if not isinstance(m, (int, float)) or m < 0 or m > 100:
                raise ValueError(f"Invalid mark: {m}. Marks must be between 0 and 100.")
        return marks

    def calculate_total(self) -> float:
        return sum(self.marks)

    def calculate_percentage(self) -> float:
        return (self.calculate_total() / 500) * 100

    def calculate_grade(self) -> str:
        pct = self.calculate_percentage()
        if pct >= 90:
            return "O"
        elif pct >= 80:
            return "E"
        elif pct >= 70:
            return "A"
        elif pct >= 60:
            return "B"
        elif pct >= 50:
            return "C"
        elif pct >= 40:
            return "D"
        return "F"

# Test
try:
    s1 = Student("Amit Sarkar", "13030825057", [85, 92, 78, 88, 90])
    print(f"Student: {s1.name} (Roll: {s1.roll_number})")
    print(f"Total Marks: {s1.calculate_total()}/500")
    print(f"Percentage : {s1.calculate_percentage():.2f}%")
    print(f"Grade      : {s1.calculate_grade()}")
except ValueError as e:
    print("Student validation error:", e)
```

```python
class InsufficientFundsError(Exception):
    pass

class InvalidTransactionError(Exception):
    pass

class BankAccount:
    def __init__(self, account_holder: str, initial_balance: float = 0.0):
        if initial_balance < 0:
            raise InvalidTransactionError("Initial balance cannot be negative.")
        self.account_holder = account_holder
        self.balance = float(initial_balance)

    def deposit(self, amount):
        try:
            val = float(amount)
            if val <= 0:
                raise InvalidTransactionError("Deposit amount must be strictly greater than zero.")
            self.balance += val
            print(f"Deposited: ₹{val:.2f} | Current Balance: ₹{self.balance:.2f}")
        except (ValueError, TypeError):
            raise InvalidTransactionError(f"Invalid input '{amount}'. Amount must be a positive number.")

    def withdraw(self, amount):
        try:
            val = float(amount)
            if val <= 0:
                raise InvalidTransactionError("Withdrawal amount must be strictly greater than zero.")
            if val > self.balance:
                raise InsufficientFundsError(
                    f"Insufficient funds: Attempted ₹{val:.2f}, balance is ₹{self.balance:.2f}."
                )
            self.balance -= val
            print(f"Withdrawn: ₹{val:.2f} | Current Balance: ₹{self.balance:.2f}")
        except (ValueError, TypeError):
            raise InvalidTransactionError(f"Invalid input '{amount}'. Amount must be a positive number.")

# Test
acc = BankAccount("Amit Sarkar", 10000.0)
try:
    acc.deposit(2500)
    acc.withdraw(5000)
    acc.withdraw(15000)   # Triggers InsufficientFundsError
except (InvalidTransactionError, InsufficientFundsError) as err:
    print("Transaction failed:", err)
```

```python
class NegativeSalaryError(Exception):
    """Raised when an entered salary is negative."""
    pass

class Employee:
    def __init__(self, emp_id: str, name: str, department: str, salary: float):
        if salary < 0:
            raise NegativeSalaryError(f"Entered salary ({salary}) cannot be negative.")
        self.emp_id = emp_id
        self.name = name
        self.department = department
        self.salary = float(salary)

    def calculate_annual_salary(self) -> float:
        return self.salary * 12

    def salary_category(self) -> str:
        annual = self.calculate_annual_salary()
        if annual >= 1500000:
            return "Executive Tier"
        elif annual >= 800000:
            return "Senior Professional Tier"
        elif annual >= 400000:
            return "Mid Tier"
        else:
            return "Associate Tier"

# Test
try:
    emp = Employee("E101", "Amit Sarkar", "AI/ML", 75000.0)
    print(f"Employee ID   : {emp.emp_id}")
    print(f"Name          : {emp.name}")
    print(f"Department    : {emp.department}")
    print(f"Monthly Salary: ₹{emp.salary:,.2f}")
    print(f"Annual Salary : ₹{emp.calculate_annual_salary():,.2f}")
    print(f"Category      : {emp.salary_category()}")
except NegativeSalaryError as err:
    print("Employee Error:", err)
```
