```python
import math
n, num, a, b, c = 5, 153, 25, 89, 42
print(f"Part A inputs loaded: n={n}, num={num}, a={a}, b={b}, c={c}")
```

```python
print(f"Factorial of {n}:", math.factorial(n))
```

```python
print(f"{num} is Prime:", num > 1 and all(num % i != 0 for i in range(2, int(num**0.5) + 1)))
```

```python
t, rev = num, 0
while t > 0: rev, t = rev * 10 + t % 10, t // 10
print(f"Reverse of {num}:", rev)
```

```python
print(f"{num} is Palindrome:", str(num) == str(num)[::-1])
```

```python
print(f"Sum of digits of {num}:", sum(map(int, str(num))))
```

```python
for i in range(1, n + 1): print(*range(1, i + 1))
```

```python
print(f"Largest among ({a}, {b}, {c}):", max(a, b, c))
```

```python
print(f"{num} is Armstrong:", num == sum(int(d)**len(str(num)) for d in str(num)))
```

```python
for i in range(1, 11): print(f"{n} x {i} = {n * i}")
```

```python
for i in range(1, n + 1): print("* " * i)
```
