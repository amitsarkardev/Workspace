```python
nums = [15, 42, 8, 93, 27, 42, 60, 15, 4, 88]
list2 = [10, 42, 50, 88, 99]
tup = tuple(nums)
print("Part B Master List loaded:", nums)
```

```python
print(f"Largest: {max(nums)}, Smallest: {min(nums)}")
```

```python
print(f"Sum: {sum(nums)}, Average: {sum(nums)/len(nums):.2f}")
```

```python
evens = sum(x % 2 == 0 for x in nums)
print(f"Even count: {evens}, Odd count: {len(nums) - evens}")
```

```python
print("Without duplicates:", list(dict.fromkeys(nums)))
```

```python
print("Reversed list:", nums[::-1])
```

```python
print(f"Tuple Max: {max(tup)}, Min: {min(tup)}")
```

```python
print("Even Tuple:", tuple(x for x in tup if x % 2 == 0))
print("Odd Tuple :", tuple(x for x in tup if x % 2 != 0))
```

```python
t = tuple(nums)
print(f"Tuple: {t}\nAverage: {sum(t)/len(t):.2f}")
```

```python
print("Second Largest:", sorted(set(nums))[-2])
```

```python
print("Merged without duplicates:", list(dict.fromkeys(nums + list2)))
```
