```python
marks = {"Amit": 85, "Rohan": 55, "Priya": 92, "Sneha": 48}
salaries = {"Amit": 85000, "Rohan": 55000, "Priya": 92000, "Sneha": 48000}
sentence = "python is simple and python is fast"
products = {"Laptop": (2, 50000), "Mouse": (5, 500), "Keyboard": (3, 1200)}
print("Part C master dictionaries loaded.")
```

```python
print("Scored > 60:", {k: v for k, v in marks.items() if v > 60})
```

```python
top = max(salaries, key=salaries.get)
print(f"Highest Salary: {top} (₹{salaries[top]})")
```

```python
words = sentence.split()
print("Word Frequency:", {w: words.count(w) for w in dict.fromkeys(words)})
```

```python
print("Squares (1-10):", {x: x**2 for x in range(1, 11)})
```

```python
d = marks.copy()
d["Vikram"], d["Rohan"] = 76, 65  # Add 'Vikram', Update 'Rohan'
d.pop("Sneha", None)              # Delete 'Sneha'
print("Modified Dictionary:", d)
```

```python
print("Total Cost: ₹", sum(qty * price for qty, price in products.values()))
```
