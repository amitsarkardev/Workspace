```python
def is_prime(n): return n > 1 and all(n % i != 0 for i in range(2, int(n**0.5) + 1))
print("Is 29 Prime?:", is_prime(29))
```

```python
def is_palindrome(v): return str(v) == str(v)[::-1]
print("121:", is_palindrome(121), "| 'madam':", is_palindrome("madam"))
```

```python
def fib(n):
    a, b, res = 0, 1, []
    for _ in range(n): res.append(a); a, b = b, a + b
    return res
print("First 8 Fibonacci terms:", fib(8))
```

```python
class Student:
    def __init__(self, name, roll, marks):
        if len(marks) != 5 or any(m < 0 or m > 100 for m in marks): raise ValueError("Provide 5 marks between 0 and 100.")
        self.name, self.roll, self.marks = name, roll, marks
    def total(self): return sum(self.marks)
    def percentage(self): return self.total() / 5
    def grade(self): return "O" if self.percentage() >= 90 else "A" if self.percentage() >= 75 else "B" if self.percentage() >= 60 else "C"

try:
    s = Student("Amit", 57, [88, 92, 85, 90, 95])
    print(f"{s.name} (Roll {s.roll}) | Total: {s.total()} | Pct: {s.percentage()}% | Grade: {s.grade()}")
    Student("Rohan", 12, [80, -10, 75, 85, 90])
except ValueError as e: print("Handled Invalid Marks:", e)
```

```python
class BankAccount:
    def __init__(self, bal=5000.0): self.bal = bal
    def deposit(self, amt):
        amt = float(amt)
        if amt <= 0: raise ValueError("Amount must be > 0.")
        self.bal += amt; print(f"Deposited ₹{amt}. Balance: ₹{self.bal}")
    def withdraw(self, amt):
        amt = float(amt)
        if amt <= 0 or amt > self.bal: raise ValueError("Invalid amount or insufficient balance.")
        self.bal -= amt; print(f"Withdrew ₹{amt}. Balance: ₹{self.bal}")

acc = BankAccount()
for fn, val in [(acc.deposit, 1500), (acc.withdraw, 2000), (acc.withdraw, 9000), (acc.deposit, -100), (acc.deposit, "xyz")]:
    try: fn(val)
    except Exception as e: print(f"Handled Error ({val}): {e}")
```

```python
class Employee:
    def __init__(self, eid, name, dept, sal):
        if sal < 0: raise ValueError("Salary cannot be negative!")
        self.eid, self.name, self.dept, self.sal = eid, name, dept, sal
    def annual_salary(self): return self.sal * 12
    def category(self): return "High" if self.sal >= 75000 else "Medium" if self.sal >= 40000 else "Low"

try:
    emp = Employee(101, "Amit", "AI/ML", 85000)
    print(f"{emp.name} ({emp.dept}) | Annual: ₹{emp.annual_salary()} | Category: {emp.category()}")
    Employee(102, "Rohan", "IT", -30000)
except ValueError as e: print("Handled Salary Error:", e)
```
