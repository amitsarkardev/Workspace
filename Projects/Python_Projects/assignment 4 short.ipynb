```python
import numpy as np
A = np.array([[10, 25, 14], [30, 42, 55], [18, 90, 7]])
B = np.array([[2, 5, 1], [3, 4, 6], [7, 2, 8]])
print("Master 3x3 matrices A and B loaded.")
```

```python
print(f"Sum: {np.sum(A)}, Mean: {np.mean(A):.2f}, Median: {np.median(A)}, Std: {np.std(A):.2f}")
```

```python
print("Transposed Matrix A:\n", A.T)
```

```python
print("Divisible by 5:", A[A % 5 == 0])
```

```python
print("Addition:\n", A + B, "\nSubtraction:\n", A - B, "\nMultiplication:\n", A * B)
```

```python
R = np.random.randint(1, 100, 10); print(f"Random: {R} | Max: {R.max()} | Min: {R.min()}")
```

```python
print("Row-wise sums:", A.sum(axis=1), "| Column-wise sums:", A.sum(axis=0))
```

```python
print("Second-largest element:", np.unique(A)[-2])
```

```python
print("Matrix Multiplication (A @ B):\n", A @ B)
```
