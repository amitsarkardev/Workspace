```python
import numpy as np, pandas as pd

marks = np.array([78, 85, 62, 90, 74])
print(f"Max: {np.max(marks)}, Min: {np.min(marks)}, Avg: {np.mean(marks):.2f}")

df1 = pd.DataFrame({"Name": ["Amit", "Rohan", "Priya", "Sneha", "Vikram"], "Marks": marks})
print("\nScored > 75:\n", df1[df1["Marks"] > 75])
print("\nHighest Marks:\n", df1.loc[df1["Marks"].idxmax()])
```

```python
import numpy as np, pandas as pd

sales = np.array([1200, 1500, 900, 1800, 2000])
print(f"Total Sales: {np.sum(sales)}, Avg Sales: {np.mean(sales):.2f}")

df2 = pd.DataFrame({"Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "SSD"], "Sales": sales})
print("\nSales > 1400:\n", df2[df2["Sales"] > 1400])
print("\nHighest Sales Product:\n", df2.loc[df2["Sales"].idxmax()])
```

```python
import numpy as np, pandas as pd

salaries = np.array([45000, 60000, 38000, 85000, 52000])
avg_sal = np.mean(salaries)
print(f"Max: {np.max(salaries)}, Min: {np.min(salaries)}, Avg: {avg_sal:.2f}")

df3 = pd.DataFrame({"Employee Name": ["Amit", "Rohan", "Priya", "Sneha", "Vikram"], "Salary": salaries})
print("\nAbove Average Salary:\n", df3[df3["Salary"] > avg_sal])
print("\nSorted by Salary (Desc):\n", df3.sort_values(by="Salary", ascending=False))
```

```python
import numpy as np, pandas as pd

ages = np.array([28, 54, 42, 65, 35])
print(f"Ages: {ages} | Avg: {np.mean(ages):.1f}, Max: {np.max(ages)}, Min: {np.min(ages)}")

df4 = pd.DataFrame({
    "Patient ID": [101, 102, 103, 104, 105],
    "Patient Name": ["Amit", "Rohan", "Priya", "Sneha", "Vikram"],
    "Age": ages,
    "Disease": ["Fever", "Cardiac", "Ortho", "Diabetes", "Flu"],
    "Treatment Cost": [4500, 25000, 14000, 18000, 3000]
})
print("\nCost > ₹10,000:\n", df4[df4["Treatment Cost"] > 10000])
print("\nAverage Treatment Cost: ₹", df4["Treatment Cost"].mean())

df4["Status"] = ["Discharged", "Admitted", "Admitted", "Admitted", "Discharged"]
print("\nPatients Age > 50:\n", df4[df4["Age"] > 50])
print("\nHighest Treatment Cost Patient:\n", df4.loc[df4["Treatment Cost"].idxmax()])

df4 = df4.sort_values(by="Treatment Cost", ascending=False)
df4.to_csv("hospital_records.csv", index=False)
print("\nSorted & Saved 'hospital_records.csv':\n", df4)
```

```python
import numpy as np, pandas as pd

tickets = np.array([150, 85, 220, 95, 180])
print(f"Total: {np.sum(tickets)}, Avg: {np.mean(tickets):.1f}, Max: {np.max(tickets)}, Min: {np.min(tickets)}")

df5 = pd.DataFrame({
    "Movie Name": ["Inception", "Avatar", "Interstellar", "Dune", "Oppenheimer"],
    "Genre": ["Sci-Fi", "Action", "Sci-Fi", "Adventure", "Drama"],
    "Ticket Price": [250, 300, 200, 280, 320],
    "Tickets Sold": tickets
})
df5["Revenue"] = df5["Ticket Price"] * df5["Tickets Sold"]
print("\nSold > 100 Tickets:\n", df5[df5["Tickets Sold"] > 100])
print("\nHighest Revenue Movie:\n", df5.loc[df5["Revenue"].idxmax()])
print(f"\nTotal Revenue: ₹{df5['Revenue'].sum()} | Avg Ticket Price: ₹{df5['Ticket Price'].mean():.2f}")

df5 = df5.sort_values(by="Revenue", ascending=False)
df5.to_csv("movie_booking.csv", index=False)
print("\nSorted & Saved 'movie_booking.csv':\n", df5)
```

```python
import numpy as np, pandas as pd

vehicles = np.array([620, 460, 800, 385, 530])
print(f"Total: {np.sum(vehicles)}, Avg: {np.mean(vehicles):.1f}, Max: {np.max(vehicles)}, Min: {np.min(vehicles)}")

df6 = pd.DataFrame({
    "Road Name": ["MG Road", "Park Street", "EM Bypass", "VIP Road", "BT Road"],
    "Location": ["Central", "South", "East", "North-East", "North"],
    "Cars": [300, 220, 400, 180, 250],
    "Buses": [70, 40, 100, 35, 60],
    "Bikes": [250, 200, 300, 170, 220]
})
df6["Total Vehicles"] = df6["Cars"] + df6["Buses"] + df6["Bikes"]
print("\nRoads with > 500 Vehicles:\n", df6[df6["Total Vehicles"] > 500])
print("\nHighest Traffic Road:\n", df6.loc[df6["Total Vehicles"].idxmax()])
print("\nLowest Traffic Road:\n", df6.loc[df6["Total Vehicles"].idxmin()])
print("\nAverage Cars Recorded:", df6["Cars"].mean())

df6 = df6.sort_values(by="Total Vehicles", ascending=False)
df6.to_csv("traffic_monitoring.csv", index=False)
print("\nSorted & Saved 'traffic_monitoring.csv':\n", df6)
```
